## 1.1 Bối cảnh Doanh nghiệp & Bài toán Kinh doanh Thực tế

### 1.1.1 Bối cảnh Doanh nghiệp & Thách thức Thị trường
Dự án nghiên cứu hành vi người tiêu dùng trên một sàn thương mại điện tử B2C đa ngành hàng, bao gồm Thời trang, Điện tử, Đồ gia dụng, Thể thao và Bách hóa. Nền tảng phục vụ khách hàng trên cả ứng dụng di động và website máy tính. Bộ dữ liệu nghiên cứu gồm 500.000 bản ghi được công bố bởi tác giả Gautham Vijayaraj trên Kaggle.

Trong bối cảnh cạnh tranh gay gắt hiện nay, doanh nghiệp đối mặt với hai thực tế lớn:
- **Áp lực chi phí thu hút khách hàng:** Báo cáo từ SimplicityDX cho thấy chi phí thu hút một khách hàng mới đã tăng tới 222% trong những năm gần đây, đặt ra áp lực tài chính rất lớn buộc doanh nghiệp phải tối ưu hóa từng đồng ngân sách tiếp thị.
- **Đặc điểm tệp khách hàng nghiên cứu:** Khác với lượng truy cập xem dạo thông thường vốn có tỷ lệ mua rất thấp (chỉ khoảng 2.0% – 2.3% theo báo cáo chuẩn ngành của IRP Commerce và Adobe), bộ dữ liệu 500.000 dòng này tập trung vào nhóm khách hàng đã đăng ký tài khoản, có thu nhập và lịch sử mua sắm định kỳ. Vì vậy, tỷ lệ giao dịch đạt mức 43.34% (tương ứng tỷ lệ không mua là 56.66% sau khi làm sạch) — phản ánh thực trạng chuyển đổi mà doanh nghiệp cần ưu tiên thấu hiểu và tối ưu hóa.

### 1.1.2 Vấn đề Cốt lõi Cần Giải Quyết
Trong kinh doanh thương mại điện tử, việc phân bổ nguồn lực tiếp thị (ngân sách quảng cáo và ưu đãi giảm giá) luôn là một bài toán tối ưu hóa giữa chi phí và doanh thu. Dữ liệu thực nghiệm trên 500.000 bản ghi cho thấy hành vi khách hàng có sự phân hóa rõ rệt: có nhóm khách hàng sẵn sàng giao dịch mà không cần mã giảm giá (`DiscountsAvailed = 0`), trong khi có nhóm nhận tới 4 – 5 voucher nhưng vẫn không phát sinh đơn hàng.

Thực tế này đặt ra bài toán kinh doanh then chốt: **Làm thế nào để phân bổ ưu đãi đúng đối tượng và đúng thời điểm?** Nếu thiếu mô hình định lượng xác suất chuyển đổi cho từng phiên tương tác, việc ra quyết định phân bổ nguồn lực sẽ đối mặt với hai rủi ro tài chính lớn:
1. **Xói mòn biên lợi nhuận (Margin Erosion):** Cấp ưu đãi giảm giá cho nhóm khách hàng vốn đã có sẵn ý định mua với mức giá niêm yết, gây sụt giảm lợi nhuận biên trên mỗi đơn hàng mà không đem lại giá trị gia tăng thực chất.
2. **Bỏ lỡ cơ hội chuyển đổi & Lãng phí chi phí tiếp cận:** Không nhận diện và kịp thời kích cầu nhóm khách hàng đang do dự (nhóm chỉ cần một tác động nhỏ là chuyển đổi thành công); đồng thời tiếp tục tiêu tốn chi phí tiếp thị vào nhóm khách hoàn toàn không có ý định mua sắm.

### 1.1.3 Mục tiêu Dự án & Kịch bản Ứng dụng Thực tế
Dự án hướng tới xây dựng mô hình dự báo **Xác suất mua hàng (Purchase Probability)** theo thời gian thực cho từng phiên tương tác, phục vụ 3 kịch bản can thiệp nghiệp vụ chính xác:
- **Kịch bản 1 — Khách hàng tiềm năng cao (Xác suất mua $\ge 70\%$):** Giữ nguyên giá bán niêm yết, hạn chế tối đa việc cấp voucher giảm giá sâu để bảo toàn biên lợi nhuận thu về.
- **Kịch bản 2 — Khách hàng do dự (Xác suất mua từ $30\%$ đến $70\%$):** Đây là phân khúc trọng tâm cần kích cầu. Hệ thống tự động kích hoạt ưu đãi kịp thời (mã miễn phí vận chuyển hoặc phiếu giảm giá ngắn hạn) để thúc đẩy quyết định chốt đơn ngay trong phiên.
- **Kịch bản 3 — Khách hàng tham khảo (Xác suất mua $< 30\%$):** Điều hướng sang kênh chăm sóc tự nhiên chi phí thấp, ngừng phân bổ ngân sách quảng cáo trả phí để chống lãng phí ngân sách tiếp thị.

Song song với mục tiêu chuyển đổi tức thì, mô hình còn hỗ trợ tối ưu hóa **Giá trị vòng đời khách hàng (Customer Lifetime Value - LTV)** thông qua việc kích thích tần suất mua sắm lặp lại và gia tăng mức độ gắn kết của hội viên thân thiết.

## 1.3 Hệ Thống 3 Câu Hỏi Nghiên Cứu Cốt Lõi (Core Research Questions)

Xuất phát từ vấn đề gốc của doanh nghiệp — **tỷ lệ không mua hàng chiếm 56.66%** (chuyển đổi chỉ đạt 43.34%) và chi phí thu hút khách hàng mới (CAC) gia tăng, đề án tập trung giải quyết 3 câu hỏi nghiên cứu trọng tâm:

* **Câu hỏi 1 (Đặc điểm Khách hàng & Nhân khẩu học):**
  > *Trong các yếu tố thuộc về đặc điểm người dùng như độ tuổi (`Age`), phân khúc (`CustomerSegment`) và mức thu nhập hàng năm (`AnnualIncome`), yếu tố nào có mối liên hệ mạnh nhất và tác động như thế nào đến quyết định mua hàng (`PurchaseStatus`)*?

* **Câu hỏi 2 (Chính sách Tiếp thị & Đòn bẩy Kích cầu):**
  > *Trong các chính sách kích cầu và kênh tiếp cận gồm kênh tiếp thị (`ReferralSource`), chương trình hội viên thân thiết (`LoyaltyProgram`) và số lượng voucher áp dụng (`DiscountsAvailed`), chính sách nào tạo ra đòn bẩy gia tăng chuyển đổi cao nhất và đâu là điểm bão hòa của ưu đãi*?

* **Câu hỏi 3 (Độ trễ Giao dịch & Tiền đề Dự báo):**
  > *Độ trễ kể từ lần giao dịch gần nhất (`LastPurchaseDaysAgo`) có mối quan hệ thuận hay nghịch với quyết định mua hàng (`PurchaseStatus`), và đâu là mốc thời gian quyết định khiến xác suất chốt đơn của khách hàng sụt giảm nghiêm trọng*?

## 1.4 Bộ Chỉ số Đánh giá Hiệu quả (Business KPIs & ML Metrics)

### 1.4.1 Chỉ số Đo lường Nghiệp vụ (Business KPIs)
Dự án tập trung vào 2 chỉ số định lượng nghiệp vụ cốt lõi làm kim chỉ nam đo lường và ra quyết định:

1. **Tỷ lệ chuyển đổi chuẩn toàn sàn (Baseline Conversion Rate - CVR):**
   $$\text{CVR} = \frac{\text{Số khách mua hàng}}{\text{Tổng số khách truy cập}} \times 100\%$$
   * *Mốc chuẩn thực nghiệm:* Đạt **43.34%** (tương ứng tỷ lệ thất thoát cơ sở là **56.66%**). Mốc 56.66% này là đối chuẩn cơ sở (Baseline Benchmark) để đánh giá năng lực cải thiện của mô hình dự báo.

2. **Mức tăng chuyển đổi biên & Ngưỡng bão hòa voucher ($\Delta\text{CVR}$):**
   $$\Delta \text{CVR}_{(k)} = \text{CVR}_{(k)} - \text{CVR}_{(k-1)}$$
   * *Ý nghĩa:* Đo lường mức tăng tỷ lệ chuyển đổi khi cấp thêm voucher thứ $k$. Khi $\Delta \text{CVR} \le 0$, ưu đãi đạt điểm bão hòa. Đây là căn cứ định lượng (tại SQL 6 và Biểu đồ 7) để thiết lập quy tắc chặn cứng trần **tối đa 5 voucher/khách** cho Kịch bản 2 ở File 6 nhằm bảo toàn biên lợi nhuận.

### 1.4.2 Chỉ số Đánh giá Kỹ thuật Mô hình (Machine Learning Metrics)
Mô hình Logistic Regression được đánh giá qua bộ 5 chỉ số gắn liền với bài toán kinh doanh, so sánh trực tiếp với mốc tham chiếu Baseline ngẫu nhiên (56.66%):
- **Accuracy (Độ chính xác tổng quát - 89.80%):** Vượt trội so với mốc Baseline 56.66%, khẳng định năng lực phân loại tổng thể đáng tin cậy.
- **Precision (Kiểm soát chi phí tiếp thị - 85.64%):** Tránh cấp nhầm voucher cho khách chỉ lướt xem dạo.
- **Recall (Khai thác tối đa doanh thu - 91.86%):** Hạn chế tối đa nguy cơ bỏ sót khách mua tiềm năng.
- **F1-Score (Cân bằng tối ưu - 88.64%):** Trung bình điều hòa giữa Precision và Recall.
- **ROC-AUC (Năng lực phân tách đa ngưỡng - 0.9658):** Đánh giá chất lượng mô hình trên toàn dải ngưỡng, là cơ sở khoa học thiết lập 2 ngưỡng xác suất $P = 0.30$ và $P = 0.70$ phân tách 3 kịch bản vận hành thực tế.

## 1.6 Đánh giá Rủi ro Rò rỉ Dữ liệu & Ma Trận Dòng Thời Gian (Temporal Leakage Audit Matrix)

Trong thực tiễn xây dựng hệ thống AI dự báo hành vi người tiêu dùng theo thời gian thực, kiểm soát **Data Leakage (Rò rỉ dữ liệu từ tương lai)** là nguyên tắc số 1 để tránh tình trạng mô hình đạt độ chính xác ảo nhưng thất bại khi vận hành. Dựa trên vòng đời tương tác của khách hàng trong một phiên, dự án phân định 17 thuộc tính theo ma trận dòng thời gian:

| Dòng thời gian thu thập | Nhóm đặc trưng dữ liệu (Features) | Mức độ rủi ro (Leakage Risk) | Giả định kiểm toán & Hướng xử lý Machine Learning |
| :--- | :--- | :---: | :--- |
| **Thời điểm T0: Đầu phiên**<br>*(Khách vừa mở app/web, chưa phát sinh hành động mới)* | • **Nhân khẩu học & Tài chính:** `Age`, `Gender`, `Region`, `AnnualIncome`<br>• **Thâm niên & Kênh dẫn nguồn:** `CustomerTenureYears`, `ReferralSource`<br>• **Thiết bị & Ngành hàng:** `PreferredDevice`, `ProductCategory` | **0% (Tuyệt đối an toàn)** | **Đặc trưng tĩnh (Static Baseline):** Dữ liệu hồ sơ đã có sẵn trên hệ thống trước khi phiên diễn ra, an toàn 100% để đưa vào mọi mô hình dự báo thời gian thực. |
| **Thời điểm T0*: Diễn tiến phiên**<br>*(Khách đang tương tác, lướt sản phẩm, áp mã)* | • **Lịch sử tích lũy:** `NumberOfPurchases`, `LastPurchaseDaysAgo`<br>• **Tương tác trong phiên:** `TimeSpentOnWebsite`, `SessionCount`<br>• **Chính sách ưu đãi:** `LoyaltyProgram`, `DiscountsAvailed` | **Rủi ro ngầm (Cần kiểm soát)** | **Giả định trạng thái (Stateful Assumption):**<br>• `NumberOfPurchases` và `LastPurchaseDaysAgo` được chốt tại mốc *trước khi phiên bắt đầu* (không bao gồm đơn hiện tại).<br>• `TimeSpentOnWebsite` và `DiscountsAvailed` đo lường tại thời điểm cắt ngang (Snapshot) của phiên. |
| **Thời điểm T1: Khoảnh khắc giao dịch**<br>*(Khách quyết định bấm thanh toán)* | • Trạng thái mua sắm: `PurchaseStatus` (0 hoặc 1) | **Biến Mục Tiêu**<br>*(Target Label)* | Nhãn phân loại nhị phân ($Y \in \{0, 1\}$) cần dự đoán. |
