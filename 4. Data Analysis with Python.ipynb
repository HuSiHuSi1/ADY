## 4.2 Thống Kê Mô Tả Nâng Cao (Descriptive Statistics: Skewness & Kurtosis)

**Câu hỏi nghiên cứu:** Các biến số định lượng phân phối như thế nào? Có bị lệch (Skewed) hay có đuôi dày (Kurtosis) không?
**Cách làm:** Tính toán đầy đủ Mean, Median, Std, Min, Max, Độ lệch (Skewness) và Độ nhọn (Kurtosis).


```python
num_cols = ['Age', 'AnnualIncome', 'NumberOfPurchases', 'TimeSpentOnWebsite', 
            'CustomerTenureYears', 'LastPurchaseDaysAgo', 'DiscountsAvailed', 'SessionCount']

desc_stats = []
for c in num_cols:
    s = df[c]
    desc_stats.append({
        'Biến số': c,
        'Trung bình (Mean)': np.round(s.mean(), 2),
        'Trung vị (Median)': np.round(s.median(), 2),
        'Độ lệch chuẩn (Std)': np.round(s.std(), 2),
        'Nhỏ nhất (Min)': np.round(s.min(), 2),
        'Lớn nhất (Max)': np.round(s.max(), 2),
        'Độ lệch (Skewness)': np.round(stats.skew(s), 3),
        'Độ nhọn (Kurtosis)': np.round(stats.kurtosis(s), 3)
    })

desc_df = pd.DataFrame(desc_stats)
display(desc_df)
```

> **Nhận xét thống kê mô tả:**
> - Các biến `Age` (Skewness = **-0.010**) và `TimeSpentOnWebsite` (Skewness = **0.019**) có độ lệch rất gần 0 (nằm trong khoảng `-0.05 đến 0.05`), cùng với `DiscountsAvailed` (**0.057**) và `NumberOfPurchases` (**-0.072**), biểu thị phân phối gần như đối xứng hoàn hảo.
> - Biến `AnnualIncome` có độ lệch phải nhẹ (Skewness = **0.178**, nhưng vẫn rất thấp so với ngưỡng cảnh báo lệch nặng $|\text{Skew}| > 1$), với giá trị trung bình (**85,018.02 USD**) và trung vị (**83,686.29 USD**) khá sát nhau.
> - Riêng `CustomerTenureYears` (**1.851**) và `SessionCount` (**1.223**) có độ lệch phải rõ rệt (đa số khách hàng có thâm niên ngắn và số phiên ít). Nhìn chung, sau khi áp dụng bộ 7 quy tắc làm sạch DAMA-DMBOK (loại bỏ 67.744 dòng lỗi), dữ liệu có độ ổn định cao và không bị méo lệch bởi outlier cực đoan.

## 4.4 Kiểm Định Giả Thuyết Thống Kê (Two-Sample Independent T-Test)

**Câu hỏi nghiên cứu:** Giá trị trung bình của 8 biến định lượng (`AnnualIncome`, `TimeSpentOnWebsite`, `NumberOfPurchases`, `Age`, `DiscountsAvailed`, `SessionCount`, `LastPurchaseDaysAgo`, `CustomerTenureYears`) ở nhóm Mua hàng (`PurchaseStatus = 1`) có sự khác biệt mang ý nghĩa thống kê so với nhóm Không mua (`PurchaseStatus = 0`) hay không?
* **Giả thuyết kiểm định (cho từng biến định lượng):**
  * $H_0$: $\mu_1 = \mu_0$ (Không có sự khác biệt về giá trị trung bình giữa nhóm mua và không mua).
  * $H_1$: $\mu_1 \neq \mu_0$ (Có sự khác biệt có ý nghĩa thống kê giữa 2 nhóm).
  * Mức ý nghĩa: $\alpha = 0.05$.


```python
group_1 = df[df['PurchaseStatus'] == 1]
group_0 = df[df['PurchaseStatus'] == 0]

test_vars = ['AnnualIncome', 'TimeSpentOnWebsite', 'NumberOfPurchases', 'Age', 'DiscountsAvailed', 'SessionCount', 'LastPurchaseDaysAgo', 'CustomerTenureYears']
ttest_results = []

for var in test_vars:
    t_stat, p_val = stats.ttest_ind(group_1[var], group_0[var], equal_var=False)
    diff = group_1[var].mean() - group_0[var].mean()
    ttest_results.append({
        'Biến kiểm định': var,
        'Mean (Mua = 1)': np.round(group_1[var].mean(), 2),
        'Mean (Không mua = 0)': np.round(group_0[var].mean(), 2),
        'Chênh lệch (Diff)': np.round(diff, 2),
        'T-Statistic': np.round(t_stat, 2),
        'P-Value': f"{p_val:.2e}",
        'Ý nghĩa thống kê (p < 0.05)': 'Bác bỏ H0 (Có ý nghĩa)' if p_val < 0.05 else 'Chấp nhận H0'
    })

ttest_df = pd.DataFrame(ttest_results)
display(ttest_df)
```

> **Kết luận kiểm định T-Test (Toàn bộ 8 biến định lượng):**
>
> - **7/8 biến có ý nghĩa thống kê cao ($p < 0.001$, Bác bỏ $H_0$):**
>   1. `LastPurchaseDaysAgo` ($t = -712.70$): Khách mua hàng vừa phát sinh giao dịch cách đây **15.48 ngày** so với **89.79 ngày** của khách không mua (chênh lệch **-74.31 ngày**). Đây là chỉ số tác động mạnh nhất đến quyết định chốt đơn.
>   2. `AnnualIncome` ($t = 60.64$): Khách mua hàng có mức thu nhập cao hơn **7,344.14 USD** (89,178.92 vs 81,834.77 USD).
>   3. `NumberOfPurchases` ($t = 44.54$): Khách mua hàng có lịch sử mua nhiều hơn **0.80 đơn** (12.04 vs 11.25 đơn).
>   4. `TimeSpentOnWebsite` ($t = 42.90$): Khách mua hàng dành nhiều thời gian trên web hơn **2.26 phút** (32.43 vs 30.17 phút).
>   5. `Age` ($t = -43.95$): Khách mua hàng có độ tuổi trung bình trẻ hơn **2.11 tuổi** (42.96 vs 45.06 tuổi).
>   6. `DiscountsAvailed` ($t = 35.91$): Khách mua hàng sử dụng nhiều hơn **0.21 mã** giảm giá (3.27 vs 3.06 mã).
>   7. `CustomerTenureYears` ($t = -26.63$): Khách mua có thâm niên trung bình là **2.20 năm** so với **2.38 năm** của khách không mua. Về toán học có ý nghĩa thống kê ($p < 0.001$), nhưng mức chênh lệch thực tế chỉ là **2 tháng** (-0.17 năm), không tạo ra tác động thực tiễn lớn trong kinh doanh.
>
> - **1/8 biến KHÔNG có ý nghĩa thống kê ($p = 0.281 > 0.05$, Chấp nhận $H_0$):**
>   - `SessionCount` ($t = 1.08, p = 0.281$): Số phiên truy cập trung bình của nhóm mua và không mua bằng nhau (**2.35 phiên**, chênh lệch **0.00**). Tần suất vào web nhiều hay ít không ảnh hưởng đến tỷ lệ chuyển đổi mua hàng.


## 4.5 Phân Tích Phương Sai Một Yếu Tố (One-Way ANOVA)

**Câu hỏi nghiên cứu:** Thời gian trên website (`TimeSpentOnWebsite`) có sự khác biệt có ý nghĩa thống kê giữa 3 phân khúc khách hàng (`Regular`, `Premium`, `VIP`) hay không?
* **Giả thuyết kiểm định:**
  * $H_0$: $\mu_1 = \mu_2 = \mu_3$.
  * $H_1$: Có ít nhất một cặp phân khúc có thời gian trung bình khác nhau.


```python
seg_groups = [group['TimeSpentOnWebsite'].values for _, group in df.groupby('CustomerSegment')]
f_stat, p_val = stats.f_oneway(*seg_groups)

print("KẾT QUẢ KIỂM ĐỊNH ONE-WAY ANOVA (TimeSpentOnWebsite theo CustomerSegment):")
print(f"F-Statistic: {f_stat:.4f}")
print(f"P-Value    : {p_val:.4e}")

for seg, grp in df.groupby('CustomerSegment'):
    print(f" - Phân khúc {seg:8s}: Mean = {grp['TimeSpentOnWebsite'].mean():.2f} phút (n = {len(grp):,})")

if p_val < 0.05:
    print(" KẾT LUẬN: Bác bỏ H0 — Thời gian truy cập website giữa các phân khúc khách hàng có sự khác biệt rõ rệt!")
else:
    print(" KẾT LUẬN: Không đủ bằng chứng bác bỏ H0.")
```

> **Kết luận kiểm định ANOVA:**
> - Với $F = 0.0244$ và $p = 0.9759 > 0.05$, ta **Chấp nhận giả thuyết $H_0$** (Không có sự khác biệt có ý nghĩa thống kê về thời gian duyệt web giữa các nhóm khách hàng).
> - Dữ liệu thực tế cho thấy cả 3 phân khúc Regular (31.16 phút), Premium (31.15 phút) và VIP (31.15 phút) đều có thời gian lướt website tương đương nhau (~31.15 phút). Phân hạng khách hàng không làm thay đổi thời gian tương tác trên web.


## 4.6 Kiểm Định Chi-Square ($\chi^2$) về Tính Độc Lập & Hệ Số Cramér's V

**Câu hỏi nghiên cứu:** Quyết định mua hàng (`PurchaseStatus`) có độc lập với toàn bộ 7 biến phân loại (`LoyaltyProgram`, `CustomerSegment`, `ProductCategory`, `Gender`, `Region`, `PreferredDevice`, `ReferralSource`) hay không? Cụ thể, 3 phân khúc `VIP`, `Premium`, `Regular` có tỷ lệ mua hàng khác nhau như thế nào?
* **Phương pháp:** Kiểm định $\chi^2$ (Chi-square test of independence) và đo độ mạnh liên kết bằng hệ số **Cramér's V**:
  $$V = \sqrt{\frac{\chi^2}{n \times \min(r-1, c-1)}}$$


```python
def cramers_v(contingency_table):
    chi2 = stats.chi2_contingency(contingency_table)[0]
    n = contingency_table.sum().sum()
    r, k = contingency_table.shape
    return np.sqrt(chi2 / (n * min(r - 1, k - 1)))

cat_test_cols = ['LoyaltyProgram', 'CustomerSegment', 'PreferredDevice', 'ReferralSource', 'Region', 'ProductCategory', 'Gender']
chi2_results = []

for col in cat_test_cols:
    contingency = pd.crosstab(df[col], df['PurchaseStatus'])
    chi2, p_val, dof, _ = stats.chi2_contingency(contingency)
    v = cramers_v(contingency)
    chi2_results.append({
        'Biến phân loại': col,
        'Bậc tự do (dof)': dof,
        'Chi-Square (χ²)': np.round(chi2, 2),
        'P-Value': f"{p_val:.2e}",
        'Cramer_V (Độ mạnh)': np.round(v, 4),
        'Mối liên hệ': 'Có ý nghĩa (Phụ thuộc)' if p_val < 0.05 else 'Độc lập'
    })

chi2_df = pd.DataFrame(chi2_results).sort_values(by='Cramer_V (Độ mạnh)', ascending=False)
display(chi2_df)

print("\nCHI TIẾT TỶ LỆ MUA HÀNG (%) THEO 2 BIẾN PHÂN LOẠI MẠNH NHẤT:")
print("\n1. Theo Phân khúc Khách hàng (CustomerSegment):")
seg_cvr = df.groupby('CustomerSegment')['PurchaseStatus'].agg(['count', 'sum', lambda x: np.round(x.mean() * 100, 2)])
seg_cvr.columns = ['Tổng khách', 'Số người mua', 'Tỷ lệ mua (%)']
display(seg_cvr.sort_values(by='Tỷ lệ mua (%)', ascending=False))

print("\n2. Theo Chương trình Thành viên (LoyaltyProgram):")
loy_cvr = df.groupby('LoyaltyProgram')['PurchaseStatus'].agg(['count', 'sum', lambda x: np.round(x.mean() * 100, 2)])
loy_cvr.columns = ['Tổng khách', 'Số người mua', 'Tỷ lệ mua (%)']
display(loy_cvr.sort_values(by='Tỷ lệ mua (%)', ascending=False))
```

> **Kết luận kiểm định Chi-Square & Cramér's V (Đủ 7 biến phân loại):**
>
> - Tất cả 7 biến phân loại đều có $p < 0.05$, bác bỏ giả thuyết độc lập ($H_0$).
> - **2 yếu tố phân loại tác động mạnh nhất:**
>   1. `LoyaltyProgram` ($\chi^2 = 3,125.90$, Cramér's V = **0.0850**): Khách có thẻ thành viên (`1`) đạt tỷ lệ mua **47.50%**, vượt trội so với nhóm không có thẻ (**39.07%**).
>   2. `CustomerSegment` ($\chi^2 = 1,422.03$, Cramér's V = **0.0574**): Kiểm định chứng minh 3 phân khúc có tỷ lệ mua hàng khác nhau có ý nghĩa thống kê — nhóm **Regular** chốt đơn cao nhất (**46.01%**), tiếp đến là **Premium (44.77%)** và thấp nhất là **VIP (39.05%)**.
> - **2 biến bổ sung làm biến kiểm soát (`ProductCategory` & `Gender`):** Đạt Cramér's V lần lượt là **0.0144** và **0.0104**, đóng góp tinh chỉnh nhẹ cho mô hình dự báo ở Bước 6.
> - **3 biến tác động cực yếu (gần bằng 0):** `Region` (0.0060), `PreferredDevice` (0.0057) và `ReferralSource` (0.0048) hầu như không tạo ra khác biệt thực chất trong tỷ lệ mua hàng.


## 4.7 Phân Tích Phân Khúc RFM (Recency & Frequency Matrix)

**Câu hỏi nghiên cứu:** Khách hàng có độ mới giao dịch gần (Recency thấp) và tần suất mua sắm cao (Frequency cao) có xác suất mua sắm đạt mức bao nhiêu %?


```python
df['RecencyGroup'] = pd.qcut(df['LastPurchaseDaysAgo'], q=3, labels=['Active (Gần)', 'Warm (Vừa)', 'Cold (Xa)'])
df['FreqGroup'] = pd.qcut(df['NumberOfPurchases'], q=3, labels=['Low Freq', 'Med Freq', 'High Freq'])

rfm_pivot = df.pivot_table(
    index='RecencyGroup',
    columns='FreqGroup',
    values='PurchaseStatus',
    aggfunc=['count', lambda x: np.round(x.mean() * 100, 2)]
)
rfm_pivot.columns = ['Count: Low', 'Count: Med', 'Count: High', 'CVR(%): Low', 'CVR(%): Med', 'CVR(%): High']
print("MA TRẬN CHUYỂN ĐỔI THEO KHUNG RFM (RECENCY & FREQUENCY):")
display(rfm_pivot)
```

> **Kết luận phân tích RFM:**
> - Nhóm khách hàng **Active + High Freq** (mới mua gần đây và mua nhiều lần) sở hữu tỷ lệ chuyển đổi cao nhất nền tảng.
> - Nhóm **Cold + Low Freq** có tỷ lệ chuyển đổi thấp nhất, cần áp dụng các chiến dịch Win-back (kích hoạt lại) với ưu đãi đặc thù.


## 4.8 Kiểm Tra Đa Cộng Tuyến (Variance Inflation Factor - VIF)

**Mục tiêu:** Kiểm tra mức độ đa cộng tuyến giữa các biến đặc trưng đầu vào trước khi huấn luyện mô hình Machine Learning ở Bước 6.
* Quy tắc: VIF < 5 biểu thị mức độ an toàn cao, không bị đa cộng tuyến.


```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Chuẩn bị ma trận biến định lượng
vif_cols = ['Age', 'AnnualIncome', 'NumberOfPurchases', 'TimeSpentOnWebsite', 
            'CustomerTenureYears', 'LastPurchaseDaysAgo', 'DiscountsAvailed', 'CustomerSatisfaction']

X_vif = df[vif_cols].copy()
X_vif['intercept'] = 1.0

vif_data = pd.DataFrame()
vif_data['Feature'] = vif_cols
vif_data['VIF'] = [np.round(variance_inflation_factor(X_vif.values, i), 3) for i in range(len(vif_cols))]
vif_data = vif_data.sort_values(by='VIF', ascending=False)

print("KẾT QUẢ KIỂM TRA ĐA CỘNG TUYẾN (VIF):")
display(vif_data)

max_vif = vif_data['VIF'].max()
print(f"\nVIF lớn nhất: {max_vif}")
if max_vif < 5.0:
    print(" XÁC NHẬN: Tất cả các biến đều có VIF < 5. Dữ liệu KHÔNG bị đa cộng tuyến, hoàn toàn sẵn sàng cho Bước 6!")
```

> **Kết luận kiểm tra Đa cộng tuyến (VIF) & Giao thức Chống Rò rỉ Dữ liệu:**
>
> 1. **Về tính độc lập giữa các biến đặc trưng (VIF < 5):** Kết quả chạy thực nghiệm cho thấy toàn bộ 8 biến định lượng đều có hệ số VIF cực thấp, dao động sát mức lý tưởng từ **1.000 đến 1.006** (cao nhất là `CustomerTenureYears = 1.006` và `LastPurchaseDaysAgo = 1.004`, xa dưới ngưỡng cảnh báo 5.0). Điều này bảo chứng ma trận dữ liệu hoàn toàn **không bị hiện tượng đa cộng tuyến**, đảm bảo trọng số hồi quy ở Bước 6 ổn định và không bị chồng lấn thông tin.
> 2. **Giao thức Chống Rò rỉ Dữ liệu (No Data Leakage Protocol — Kết nối RQ5 & Mục 1.6):**
>    * Mặc dù biến `CustomerSatisfaction` đạt chỉ số VIF tuyệt đối an toàn (**1.000** $< 5$), nhưng theo thiết kế dòng thời gian giao dịch tại Bước 1, đây là biến đo lường trải nghiệm **hậu giao dịch** (Post-purchase / T2).
>    * Vì vậy, để đảm bảo mô hình dự báo hoạt động chính xác theo thời gian thực (trước khi khách quyết định bấm Mua), biến `CustomerSatisfaction` sẽ **được loại bỏ hoàn toàn khỏi tập đặc trưng huấn luyện ở Bước 6** nhằm triệt tiêu rủi ro Look-ahead Leakage.
