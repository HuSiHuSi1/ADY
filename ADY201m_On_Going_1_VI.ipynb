Đồng thời, mô hình giúp gia tăng giá trị vòng đời khách hàng qua việc thúc đẩy tần suất mua lặp lại và gắn kết hội viên thân thiết.

#### NHIỆM VỤ SINH VIÊN 1.3: XÂY DỰNG BÀI TOÁN KHOA HỌC DỮ LIỆU
*(Mô tả chi tiết bài toán dưới góc nhìn Khoa học dữ liệu)*

**BÀI LÀM CỦA SINH VIÊN:**

- **Loại bài toán:** Học có giám sát (Supervised Learning) — Phân loại nhị phân (Binary Classification).
- **Biến mục tiêu ($y$):** `PurchaseStatus` (1 = Khách hàng thực hiện mua hàng, 0 = Khách hàng không mua hàng trong phiên).
- **Các biến đặc trưng dự kiến ($X$):** 16 thuộc tính hành vi và nhân khẩu học của khách hàng, được chuẩn hóa thành 6 nhóm nghiệp vụ:
  1. *Nhân khẩu học:* `Age`, `Gender`, `Region`.
  2. *Năng lực tài chính:* `AnnualIncome`.
  3. *Lịch sử mua sắm:* `NumberOfPurchases`, `LastPurchaseDaysAgo`.
  4. *Mức độ tương tác & Sở thích:* `TimeSpentOnWebsite`, `SessionCount`, `PreferredDevice`, `ProductCategory`.
  5. *Tiếp thị & Quan hệ khách hàng:* `CustomerTenureYears`, `CustomerSegment`, `LoyaltyProgram`, `ReferralSource`, `DiscountsAvailed`.
  6. *Trải nghiệm ứng dụng:* `CustomerSatisfaction` (Khảo sát In-app CSAT ghi nhận ở cuối phiên, được kiểm soát chặt chẽ qua Khung đánh giá kép Kịch bản A/B để loại bỏ rủi ro rò rỉ dữ liệu Look-ahead Bias).
- **Chỉ số đánh giá mô hình:**
  - *Accuracy (Độ chính xác tổng thể):* Đo lường tỷ lệ dự đoán đúng trên toàn bộ tập kiểm tra, dùng làm mốc tham chiếu cơ sở.
  - *Precision (Kiểm soát chi phí tiếp thị):* Ưu tiên hàng đầu để hạn chế tối đa việc phát voucher nhầm cho khách hàng chỉ xem dạo (giảm thiểu False Positive).
  - *Recall (Khai thác tối đa doanh thu):* Đảm bảo không bỏ sót khách hàng tiềm năng thực sự muốn mua (giảm thiểu False Negative).
  - *F1-Score (Chỉ số quyết định lựa chọn thuật toán):* Cân bằng tối ưu giữa Precision và Recall trong bối cảnh chi phí CAC tăng cao.

#### NHIỆM VỤ SINH VIÊN 3.3: ÉP KIỂU DỮ LIỆU & CHUẨN HÓA ĐỊNH DẠNG
*Ép đúng kiểu dữ liệu cho từng thuộc tính và làm sạch định dạng văn bản/chuỗi.*

```python
# 1. Chuẩn hóa chuỗi ký tự: Xóa khoảng trắng thừa đầu/cuối của các cột phân loại
cat_cols = ['Gender', 'ProductCategory', 'PreferredDevice', 'Region', 'ReferralSource', 'CustomerSegment']
for col in cat_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip()

# 2. Tối ưu hóa bộ nhớ: Ép kiểu biến phân loại sang 'category' và biến nhị phân sang 'int8'
df_clean[cat_cols] = df_clean[cat_cols].astype('category')
df_clean['LoyaltyProgram'] = df_clean['LoyaltyProgram'].astype('int8')
df_clean['PurchaseStatus'] = df_clean['PurchaseStatus'].astype('int8')

# 3. Kiểm tra thông tin bộ nhớ và cấu trúc kiểu dữ liệu sau chuẩn hóa
mem_before = df_raw.memory_usage().sum() / (1024**2)
mem_after = df_clean.memory_usage().sum() / (1024**2)
print(f"Bộ nhớ trước khi tối ưu: {mem_before:.2f} MB")
print(f"Bộ nhớ sau khi tối ưu : {mem_after:.2f} MB (Tiết kiệm {(1 - mem_after/mem_before)*100:.1f}%)")
df_clean.dtypes.to_frame('Kiểu dữ liệu sau tối ưu')
```

**NHẬN XÉT CỦA SINH VIÊN (Bước 3.3):**

- **Chuẩn hóa chuỗi văn bản:** Đã thực hiện `.str.strip()` trên toàn bộ 6 thuộc tính phân loại (`Gender`, `ProductCategory`, `PreferredDevice`, `Region`, `ReferralSource`, `CustomerSegment`) nhằm loại bỏ các ký tự khoảng trắng thừa đầu/cuối (Trimming Whitespace). Thao tác này đảm bảo tính đồng nhất của các nhãn danh mục, ngăn ngừa hiện tượng phân mảnh nhãn (ví dụ: nhầm lẫn `'VIP '` với `'VIP'`) mà vẫn bảo toàn chữ hoa viết tắt chuẩn của hội viên VIP.
- **Tối ưu hóa tài nguyên:** Chuyển đổi 6 thuộc tính phân loại sang kiểu dữ liệu `category` và 2 biến nhị phân (`LoyaltyProgram`, `PurchaseStatus`) sang `int8`, giúp tiết kiệm hơn 50% dung lượng RAM tiêu thụ, tối ưu hóa tốc độ tính toán khi huấn luyện mô hình phân loại.
- **Thuộc tính thời gian:** Dữ liệu không chứa trường chuỗi ngày tháng thô (như `YYYY-MM-DD`), mà hệ thống đã tiền xử lý thành các biến định lượng liên tục (`CustomerTenureYears` đo bằng năm, `LastPurchaseDaysAgo` đo bằng ngày), do đó việc duy trì kiểu số thực `float64` và số nguyên `int64` là hoàn toàn chính xác.

#### NHIỆM VỤ SINH VIÊN 3.4: PHÁT HIỆN & XỬ LÝ NGOẠI LAI BAN ĐẦU
*Vẽ Boxplot cho các biến số quan trọng, tính ngưỡng IQR và xử lý ngoại lai (Cắt bỏ/Giới hạn giá trị).*

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Trực quan hóa nhận diện ngoại lai ban đầu bằng Boxplot
num_cols = ['Age', 'AnnualIncome', 'NumberOfPurchases', 'TimeSpentOnWebsite', 'CustomerTenureYears', 'LastPurchaseDaysAgo']
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for i, col in enumerate(num_cols):
    ax = axes[i // 3, i % 3]
    sns.boxplot(y=df_clean[col], ax=ax, color='#4A90E2')
    ax.set_title(f'Phân phối Boxplot: {col}', fontsize=11, fontweight='bold')
    ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# 2. Xử lý ngoại lai logic nghiệp vụ theo Khung 7 vấn đề kiểm toán (Audit Trail)
issues = pd.DataFrame(index=df_clean.index)
issues['neg_values'] = (df_clean['NumberOfPurchases'] < 0) | (df_clean['TimeSpentOnWebsite'] < 0) | (df_clean['CustomerTenureYears'] < 0) | (df_clean['LastPurchaseDaysAgo'] < 0)
issues['purchase_before_tenure'] = df_clean['LastPurchaseDaysAgo'] > (df_clean['CustomerTenureYears'] * 365.25)
issues['zero_purchases_has_last_date'] = (df_clean['NumberOfPurchases'] == 0) & (df_clean['LastPurchaseDaysAgo'] > 0)
issues['zero_purchases_used_discounts'] = (df_clean['NumberOfPurchases'] == 0) & (df_clean['DiscountsAvailed'] > 0)
issues['zero_purchases_vip_segment'] = (df_clean['NumberOfPurchases'] == 0) & (df_clean['CustomerSegment'].isin(['VIP', 'Premium']))
issues['underage_account_creation'] = (df_clean['Age'] - df_clean['CustomerTenureYears']) < 13
issues['bot_extreme_review'] = (df_clean['TimeSpentOnWebsite'] < 1) & (df_clean['PurchaseStatus'] == 0) & (df_clean['CustomerSatisfaction'].isin([1, 5]))

rejected_mask = issues.any(axis=1)
df_clean = df_clean[~rejected_mask].copy()

print(f"Tổng số dòng ngoại lai/lỗi bị bóc tách: {rejected_mask.sum():,} dòng ({rejected_mask.sum()/len(df_raw)*100:.2f}%)")
print(f"Tổng số dòng dữ liệu sạch hợp lệ giữ lại : {len(df_clean):,} dòng ({len(df_clean)/len(df_raw)*100:.2f}%)")
```

**NHẬN XÉT & GIẢI THÍCH CÁCH XỬ LÝ NGOẠI LAI CỦA SINH VIÊN (Bước 3.4):**

- **Các thuộc tính chứa ngoại lai & sai lệch:**
  1. `NumberOfPurchases`, `TimeSpentOnWebsite`, `CustomerTenureYears`, `LastPurchaseDaysAgo`: Xuất hiện các giá trị âm mang mã lỗi kỹ thuật (`-1`, clock drift).
  2. `LastPurchaseDaysAgo` so với `CustomerTenureYears`: Xuất hiện các giao dịch trước ngày tài khoản được tạo.
  3. `NumberOfPurchases` so với `LastPurchaseDaysAgo`, `DiscountsAvailed`, `CustomerSegment`: Mâu thuẫn khi tài khoản chưa từng mua hàng (`NumberOfPurchases == 0`) nhưng lại có ngày mua gần nhất, đã tiêu dùng voucher, hoặc được xếp hạng VIP/Premium.
  4. `Age` so với `CustomerTenureYears`: Xuất hiện tài khoản được tạo khi người dùng dưới 13 tuổi (vi phạm luật COPPA/GDPR).
  5. `TimeSpentOnWebsite` và `CustomerSatisfaction`: Xuất hiện tài khoản bot/spam truy cập dưới 1 phút, không mua hàng nhưng đánh giá phân cực cực đoan 1 sao hoặc 5 sao.

- **Số lượng ngoại lai phát hiện được:** Tổng hợp qua bảng lưu vết kiểm toán (Audit Trail) từ 7 vấn đề nghiệp vụ, nhóm đã bóc tách được **67.744 dòng vi phạm (chiếm 13,55% dữ liệu gốc)**.

- **Phương pháp xử lý (Trimming vs Capping) & Cơ sở lý luận thực tế:**
  - **Không áp dụng Trimming/Capping máy móc theo chuẩn IQR ($1.5 \times IQR$):** Trong dữ liệu thương mại điện tử, các giá trị lớn ở biên trên như thu nhập cao (`AnnualIncome > $140,000`) hay thời gian lướt web lâu (`TimeSpentOnWebsite > 50` phút) phản ánh hành vi mua sắm hoàn toàn tự nhiên của phân khúc khách hàng giàu có hoặc khách hàng kỹ tính khảo sát sản phẩm. Nếu máy móc áp dụng Winsorization/Capping theo IQR, ta sẽ vô tình bóp méo hành vi của nhóm khách hàng mang lại giá trị cao nhất cho doanh nghiệp.
  - **Áp dụng Trimming dựa trên Logic nghiệp vụ (Domain-specific Business Rules):** Nhóm chỉ loại bỏ các quan sát vi phạm nghiêm trọng tính logic và pháp lý (7 vấn đề trên). Toàn bộ 67.744 dòng này được lưu riêng vào tệp `customerData_rejected.csv` để đội ngũ kỹ thuật rà soát lỗi hệ thống, đồng thời giữ lại nguyên vẹn **432.256 dòng dữ liệu sạch hợp lệ (86,45%)**.
